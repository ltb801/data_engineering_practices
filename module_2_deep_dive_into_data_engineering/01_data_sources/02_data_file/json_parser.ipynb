1. Simple Structure

```python
import os
import json
import pandas as pd

dest_path = "destination/json"
os.makedirs(dest_path, exist_ok=True)
```

```python
# HINTS
# 1. Use pandas.read_json()
# 2. Export to CSV as path: 'destination/json'
simple_json_file_path = "simple_structure.json"

data = pd.read_json(simple_json_file_path)
print(data)
data.to_csv(f"{dest_path}/simple_structure.csv", index=False)
```

2. Multiple Level

```python
# HINTS
# 1. Use pandas json_normalize to read data from record_path
# 2. Export to CSV as path: 'destination/json'

multiple_level_json_file_path = "multiple_level.json"

with open(multiple_level_json_file_path) as f:
    data = json.load(f)

print(data)
    
df = pd.json_normalize(data, record_path=["students"], meta=["school_name", "class"])
print(df)

df.to_csv(f"{dest_path}/multiple_level.csv", index=False)
```
