1. Simple Structure

```python
import os
import json
import pandas as pd

dest_path = "destination/json"
os.makedirs(dest_path, exist_ok=True)
```

```python
# 1. Use pandas.read_json()
# 2. Export to CSV as path: 'destination/json'
simple_json_file_path = "simple_structure.json"

data = pd.read_json(simple_json_file_path)
print(data)
data.to_csv(f"{dest_path}/simple_structure.csv", index=False)
```

2. Multiple Level

```python
# 1. Use pandas json_normalize to read data from record_path
# 2. Export to CSV as path: 'destination/json'

multiple_level_json_file_path = "multiple_level.json"

with open(multiple_level_json_file_path) as f:
    data = json.load(f)

print(data)
    
df = pd.json_normalize(data, record_path=["students"], meta=["school_name", "class"], record_prefix="students_")
print(df)

df.to_csv(f"{dest_path}/multiple_level.csv", index=False)
```

3. Nested Levels

```python
# Approach 1: use record_path + meta
# 1. Open the load JSON into dictionary 
# 2. Use pandas json_normalize to read data from record_path
# 3. Export to CSV as path: 'destination/json'

nested_level_json_file_path = "nested_level.json"

with open(nested_level_json_file_path) as f:
    data = json.load(f)

print(data)
    
df = pd.json_normalize(data, 
                       record_path=["students"], 
                       meta=[
                           "school_name",
                           ["class", "level"],
                           ["class", "room"],
                           ["class", "teacher"]
                       ])
print(df)

df.to_csv(f"{dest_path}/nested_level.csv", index=False)
```

```python
# Approach 2: Read separately and cross join
# 1. Open the load JSON into dictionary 
# 2. Use pandas json_normalize to read data from record_path
# 3. Export to CSV as path: 'destination/json'

nested_level_json_file_path = "nested_level.json"

with open(nested_level_json_file_path) as f:
    data = json.load(f)

print(data)

df1 = pd.json_normalize(data=data)

df2 = pd.json_normalize(data=data['students'])

# Cross join two tables
df = pd.merge(left=df1, right=df2, how='cross')
df = df.drop(columns='students')
df.to_csv(f"{dest_path}/nested_level_advanced.csv", index=True)
```
