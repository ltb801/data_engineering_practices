Data File - Excel

```python
# input variables
excel_file_path = "auditor_and_liquidator.xlsx"
auditor_sheetname = "REG_AUDITOR_202511"
liquidator_sheetname = "LIQUIDATOR_202510"
```

```python
import os
import pandas as pd

os.makedirs("destination/excel", exist_ok=True)
```

Read Auditor data

```python
# 1. Read data from "REG_AUDITOR_202511" spreadsheet using Pandas
# 2. Apply data type conversion and export to CSV
# 3. Export to path 'destination/excel'

# Read data from spreadsheet
df = pd.read_excel(excel_file_path, sheet_name=auditor_sheetname, skiprows=2)
print(df.head())
```

```python
# Inspect data types
print(df.dtypes)
```

```python
# Enforce schema to correct data types
auditor_schema = {
    "REGISTER_NAME": str,
    "REG_AUD_NUM": str,
    "REG_AUD_NAME": str, 
    "REG_AUD_ACN": str,
    "LIQ_START_DT": 'datetime64[ns]',
    "REG_AUD_STATUS": str,
    "REG_AUD_SUSP_DT": str,
    "REG_AUD_ADD_LOCAL": str,
    "REG_AUD_ADD_STATE": str,
    "REG_AUD_ADD_PCODE": str,
    "REG_AUD_ADD_COUNTRY": str
}

df_converted = pd.read_excel(excel_file_path, sheet_name=auditor_sheetname, skiprows=2, dtype=auditor_schema)
```

```python
# Export to CSV
df_converted.to_csv(f"destination/excel/{auditor_sheetname}.csv", index=False)
```

Read Liquidator data

```python
# 1. Read data from "LIQUIDATOR_202510" spreadsheet using Pandas
#    + Provide a schema when reading data file

liquidator_schema = {
    "REGISTER_NAME": str,
    "LIQ_NUM": str,
    "OFF_LIQ_NUM": str, 
    "LIQ_NAME": str,
    "LIQ_START_DT": 'datetime64[ns]',
    "OFF_LIQ_START_DT": 'datetime64[ns]',
    "LIQ_STATUS": str,
    "LIQ_SUSP_DT": 'datetime64[ns]',
    "LIQ_ADD_LOCAL": str,
    "LIQ_ADD_STATE": str,
    "LIQ_ADD_PCODE": str,
    "LIQ_ADD_COUNTRY": str,
    "LIQ_FIRM": str
}

# 2. Export to CSV as path 'destination/excel'

df = pd.read_excel(excel_file_path, sheet_name=liquidator_sheetname, dtype=liquidator_schema)
df.to_csv(f"destination/excel/{liquidator_sheetname}.csv", index=False)
```
